```python
import os
import matplotlib.pyplot as plt
from PIL import Image

base_path = r"C:\Users\ishan\OneDrive\College\Projec\dataset"

def analyze():
    train_path = os.path.join(base_path, 'train')
    classes = ['bird', 'drone']
    
    counts = {}
    for cls in classes:
        counts[cls] = len(os.listdir(os.path.join(train_path, cls)))
    
    print(f"Class Distribution: {counts}")
    
    plt.figure(figsize=(10, 5))
    for i, cls in enumerate(classes):
        class_folder = os.path.join(train_path, cls)
        img_name = os.listdir(class_folder)[0] 
        img_path = os.path.join(class_folder, img_name)
        
        img = Image.open(img_path)
        plt.subplot(1, 2, i+1)
        plt.imshow(img)
        plt.title(f"Sample: {cls}\nSize: {img.size}")
        plt.axis('off')
    
    print("Opening plot window... (Close it to continue)")
    plt.show()

if __name__ == "__main__":
    analyze()
```
